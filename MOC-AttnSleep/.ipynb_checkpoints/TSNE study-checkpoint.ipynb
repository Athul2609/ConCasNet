## After two path CNN ##

```python
def tsne_printer(x, labels):
    # Initialize t-SNE with default parameters
    tsne = TSNE(n_components=2, random_state=42)

    # Fit and transform the data to 2D
    tsne_result_before = tsne.fit_transform(x)

    plt.figure(figsize=(10, 8))
    
    # Map colors based on class labels
    colors1 = ['grey' if label == 0 else (0.798216, 0.280197, 0.469538, 1.0) for label in labels]

    scatter = plt.scatter(tsne_result_before[:, 0], tsne_result_before[:, 1], c=colors1, alpha=0.7)
    print(colors1)
    # Add a legend
    legend_classes = [0,1]
    d = {0: "Rest-N3-Awake", 1: "N2"}
    legend_handles = [plt.Line2D([0], [0], marker='o', color='w', markerfacecolor=["grey",(0.798216, 0.280197, 0.469538, 1.0)],
                            label=d[c]) for c in legend_classes]
    
    plt.legend(handles=legend_handles, title='Classes', loc='best')
    plt.title('MOCAttnSleep N2')
    plt.xlabel('Dimension 1')
    plt.ylabel('Dimension 2')
    plt.show()
```

```python
from scipy.signal import hilbert
import scipy

def emd(signal):
    max_imf = 10  # Set the maximum number of IMFs to extract
    decomposition = []
    residual = signal.copy()

    for _ in range(max_imf):
        imf = signal - residual
        residual = imf
        decomposition.append(imf)

        # Check for monotonicity in the signal
        if np.any(np.diff(imf) > 0) and np.any(np.diff(imf) < 0):
            break

    return decomposition, residual

# Perform EMD to obtain the IMFs and residue
imfs, residue = emd(samp)
```

```python
def compute_instantaneous_frequency(imfs, fs):
    instantaneous_frequencies = []
    for imf in imfs:
        hilbert_transform = hilbert(imf)
        instantaneous_frequency = np.diff(np.unwrap(np.angle(hilbert_transform))) / (2 * np.pi) * fs
        instantaneous_frequencies.append(instantaneous_frequency)
    return instantaneous_frequencies

# Replace fs with the sampling frequency of your time-domain signal (in Hz)
fs = 100  # Example: 1000 Hz
instantaneous_frequencies = compute_instantaneous_frequency(imfs, fs)
```

```python
import matplotlib.pyplot as plt

# Plot the original time-domain signal
plt.figure(figsize=(12, 6))
plt.subplot(len(imfs)+1, 1, 1)
plt.plot(np.arange(len(samp)) / fs, samp, 'b')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude')
plt.title('Original Time-Domain Signal')

# Plot the IMFs and their instantaneous frequencies
for i in range(len(imfs)):
    plt.subplot(len(imfs)+1, 1, i+2)
    plt.plot(np.arange(len(imfs[i])) / fs, imfs[i], 'g')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Amplitude')
    plt.title(f'IMF {i+1}')
    plt.subplot(len(imfs)+1, 1, len(imfs)+2)
    plt.plot(np.arange(len(instantaneous_frequencies[i])) / fs, instantaneous_frequencies[i])
    plt.xlabel('Time (seconds)')
    plt.ylabel('Frequency (Hz)')
    plt.title(f'Instantaneous Frequency of IMF {i+1}')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import spectrogram

# Assuming your EEG signal is stored in the 'eeg_signal' variable.
# Sample frequency (sampling rate) is 100 Hz.

# Define window size and overlap for STFT.
window_size = 2  # in seconds
overlap = 0.5     # 50% overlap between frames

# Calculate the number of samples in each window and overlap.
window_size_samples = int(window_size * 100)
overlap_samples = int(window_size_samples * overlap)

# Compute the spectrogram using the STFT function from scipy.signal.
frequencies, times, Sxx = spectrogram(
    eeg_signal, fs=100, window='hamming', nperseg=window_size_samples, noverlap=overlap_samples
)

# Plot the spectrogram.
plt.figure(figsize=(10, 6))
plt.pcolormesh(times, frequencies, 10 * np.log10(Sxx), cmap='magma' ,shading='auto')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.colorbar(label='Power Spectral Density (dB)')
plt.title('EEG Spectrogram')
plt.tight_layout()
plt.show()
```
